## A. Violations Ending


```python
DATA_PATH = pathlib.Path(
            "/Users/jessbreda/Desktop/github/animal-learning/data/raw"
        )
F_NAME = "rat_behavior.csv"

RAT_DF = pd.read_csv(DATA_PATH / F_NAME)
RAT_DF["violation"] = RAT_DF.hit.isnull()
```

```python
example_animals = ['W065', 'W068', "W089" ]
example_animals_df = RAT_DF[RAT_DF.subject_id.isin(example_animals)]
example_animals_df.head()

example_animals_df = RAT_DF.query("subject_id in @example_animals").groupby(["subject_id", "session"]).agg(
    hit_rate=("hit", "mean"),
    viol_rate=("violation", "mean"),
    stage=("training_stage", "max"),

).reset_index()

example_animals_df.head()
```

```python
pal = sns.color_palette("gray", 4)
fig, ax = plt.subplots(1, 3, figsize=(20, 5), sharey=True)
for i, (animal, df) in enumerate(example_animals_df.groupby("subject_id")):
    # Apply a rolling mean with a window size of 5 sessions
    df['hit_smoothed'] = df['hit_rate'].rolling(window=21, min_periods=1).mean()
    df['violation_smoothed'] = df['viol_rate'].rolling(window=21, min_periods=1).mean()
    
    sns.lineplot(x="session", y="hit_smoothed", data=df, color=c.HIT_COLOR, ax=ax[i], errorbar=None, label="Hit")
    sns.lineplot(x="session", y="violation_smoothed", data=df, color=c.VIOLATION_COLOR, ax=ax[i], errorbar=None, label="Violation")


    # Plot stage information
    for stage in df['stage'].unique():
        if stage == 2:
            continue
        stage_df = df[df['stage'] == stage]
        ax[i].axvspan(
            xmin=stage_df['session'].min() - 0.5,
            xmax=stage_df['session'].max() + 0.5,
            ymin=0.9,
            ymax=1,
            alpha=0.3,
            color=pal[stage-1]
        )
        ax[i].text(
            x=stage_df['session'].mean(),
            y=1.13,  # slightly above the plot
            s=f"{int(stage)}",
            ha="center",
            va="center"
        )



    ax[i].set(ylabel="Performance", title=animal, xlabel="Session", ylim= (-0.1, 1.2))
    
    if i !=0:
        # remove legend
        ax[i].get_legend().remove()

fig.savefig("figures/figS1_example_animals_violation_drop.png", dpi=300, bbox_inches="tight")
```

## C-E Alignment


```python
alignment_path = "../../data/processed/dataset_alignment/"
summary = PostAlignmentVisualizer(ANIMAL_IDS, alignment_df_path=alignment_path)
summary.alignment_df.head()
```

```python

def plot_mean_line(dd, column, color, ax):
    mean_value = dd[column].mean()
    ax.axhline(mean_value, color=color, linestyle="--", label=f"Mean: {mean_value:.2f}")
    handles, labels = ax.get_legend_handles_labels()
    ax.legend(handles[:1], labels[:1], frameon=False)
```

```python
fig, ax = plt.subplots(figsize=(10, 3))

dd = summary.compute_and_plot_delta(
    "new",
    "old",
    "n_sessions",
    color=c.N_SESSION_COLOR,
    title="$\Delta$ Number of Sessions From New to Old",
    ax=ax,
)

plot_mean_line(dd, 'delta_n_sessions', c.N_SESSION_COLOR, ax)

plt.savefig(f"figures/figS1_delta_n_sessions.png", dpi=300, bbox_inches="tight")
```

```python
fig, ax = plt.subplots(figsize=(10, 3))
dd = summary.compute_and_plot_delta(
    "new_no_touts",
    "old",
    "n_trials_per_session",
    color=c.N_TRIAL_COLOR,
    title="$\Delta$ Number of Trials/Session From New to Old",
    ax=ax,
)
plot_mean_line(dd, 'delta_n_trials_per_session', c.N_TRIAL_COLOR, ax)

plt.savefig(f"figures/figS1_delta_n_trials_per_session.png", dpi=300, bbox_inches="tight")
```

# Goal: load the old data and the new data (with and without touts)

- compute the total trials and sessions in the new data
- compute the % of trials/session that were touts


```python
from violmulti.data.dataset_loader import DatasetLoader
```

```python
new_all = DatasetLoader(data_type="new_all").load_data()
new_match_old_viols = DatasetLoader(data_type="new_match_old_viols").load_data()
old_viols = DatasetLoader(data_type="old_viols").load_data()
```

```python
old_stats = old_viols.groupby(["animal_id", "session"]).violation.mean().agg(
    mean_violation = "mean",
    std_violation = "std",
).reset_index()
old_stats
```

```python
new_all_stats = new_all.groupby(["animal_id", "session"]).violation.mean().agg(
    mean_violation = "mean",
  std_violation = "std",
).reset_index()
new_all_stats
```

```python

new_match_old_viols_stats = new_match_old_viols.groupby(["animal_id", "session"]).violation.mean().agg(
    mean_violation = "mean",
  std_violation = "std",
).reset_index()
new_match_old_viols_stats
```

```python
# compute the difference between the new and old data
new_all_stats["data_type"] = "new_all"
new_match_old_viols_stats["data_type"] = "new_match_old_viols"
old_stats["data_type"] = "old"

combined_stats = pd.concat([old_stats, new_all_stats, new_match_old_viols_stats], axis=0)
combined_stats
```
