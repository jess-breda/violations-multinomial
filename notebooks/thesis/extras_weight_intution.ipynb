# Models


```python
import seaborn as sns 
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

import config as c
sns.set_context("talk")
```

# B weight intuition


```python
feature = [
    "Stimulus A",
    "Stimulus B",
]
cls = ["R-L", "R-L"]
weight = [5, -7]

binary_palette = ["purple", "purple"]

binary_df = pd.DataFrame(
    {"Feature": feature, "Weight": weight, "Class": cls, "Color": binary_palette}
)
```

```python
feature = [
    "Stimulus A",
    "Stimulus A",
    "Stimulus B",
    "Stimulus B",
]
weight = [-2.7, 2.3, 3.6, -3.4]
cls = ["L", "R", "L", "R"]

multi_palette = [
    c.LEFT_COLOR,
    c.RIGHT_COLOR,
    c.LEFT_COLOR,
    c.RIGHT_COLOR,
]

multi_df = pd.DataFrame(
    {"Feature": feature, "Weight": weight, "Color": multi_palette, "Class": cls}
)

```

```python

# Plot for Binary-class Features
fig, ax = plt.subplots(figsize=(5, 5), constrained_layout=True)

sns.barplot(
    data=binary_df,
    x="Weight",
    y="Feature",
    palette=binary_palette,
    fill=False,
    width=0.5,
    orient='h',
    label="R-L"
)

ax.axvline(0, color="k")
ax.set_xlim(-8, 6)
_ = ax.set_title("Examaple Binary Features")
sns.despine()

handles, labels = ax.get_legend_handles_labels()
ax.legend(title="Class", handles=[handles[0]], labels=[labels[0]], loc="upper left")
plt.savefig("figures/fig2_models_binary.png", bbox_inches="tight", dpi=300)
```

```python


# Plot for Multi-class Features
fig, ax = plt.subplots(figsize=(5, 5), constrained_layout=True)

sns.barplot(
    data=multi_df,
    x="Weight",
    y="Feature",
    hue="Class",
    palette=multi_palette,
    fill=False,
    orient='h'
)

ax.axvline(0, color="k")
ax.set_xlim(-8, 6)
_ = ax.set_title("Example Multi Features")
sns.despine()
ax.legend(loc="upper left")


plt.savefig("figures/fig2_models_multi.png", bbox_inches="tight", dpi=300)
```
